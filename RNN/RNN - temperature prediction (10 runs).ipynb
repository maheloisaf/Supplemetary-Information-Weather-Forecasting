```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Dados reais fornecidos
dados_reais = np.array([[26.84], [26.69], [27.41], [27.24], [26.86], 
                        [27.23], [27.33], [27.98], [28.44], [27.73], 
                        [26.42], [27.26], [27.41], [26.88]])

# Definindo a cor cornflowerblue
cor = '#18d17b'

# Criando o gráfico de violinos
plt.figure(figsize=(8, 6))

# Criando o gráfico de violinos com a cor cornflowerblue fixa
violino = sns.violinplot(
    x=dataset['mes2'], 
    y=dataset['completo'], 
    density_norm="count",
    hue=None,  # Removendo a legenda de cores
    color=cor,  # Usando cornflowerblue para todos os violinos
    saturation=1.5,  # Aumentando a saturação para uma cor mais intensa
    ax=None
)

# Ajustando os dados reais para começar em x=1
x_values = np.arange(0, len(dados_reais))

# Adicionando a curva dos dados reais (agora com x começando de 1)
plt.plot(x_values, dados_reais, color='black', marker='o', label='Dados Reais', linewidth=2)

# Configurações de rótulos, título e limites do gráfico
plt.xlabel('Tempo (dias)', fontsize=16)
plt.ylabel('Temperatura (°C)', fontsize=16)
plt.axis([-1, 14, 24.5, 30.5])  # Ajustando proporção do eixo Y
plt.title("Distribuição da previsão clássica da temperatura", fontsize=18)

# Adicionando a legenda
plt.legend()

# Ativando o fundo quadriculado com linhas claras
plt.grid(True, color='lightgray', alpha=0.5)  # Linha tracejada e cor clara

# Salvando o gráfico
plt.savefig('qdist_prevtemp')  # Trocar nome da figura, se necessário

# Exibindo o gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Dados fornecidos para os dados reais
dados_reais = np.array([[26.84], [26.69], [27.41], [27.24], [26.86], 
                        [27.23], [27.33], [27.98], [28.44], [27.73], 
                        [26.42], [27.26], [27.41], [26.88]])

# Meses (assumindo que sejam de 1 a 14)
mes = np.arange(1, len(dados_reais) + 1)  # Ajustado para começar de 1 até 14

Uu_values = np.array([Uu01, Uu02, Uu03, Uu04, Uu05, Uu06, Uu07, Uu08, Uu09, Uu10]) 

A = np.mean(Uu_values, axis=0)  # Calcula a média ao longo da primeira dimensão (média de Uu01 a Uu10)

# Dados reais para o gráfico
Teste = dados_reais.flatten()  # Convertendo para 1D, caso contrário pode gerar erro
plt.figure(figsize=(8, 6))
# Criando o gráfico com as linhas para os dados reais e as previsões
plt.plot(mes, Teste, 'k-', label='Dados Reais')  # Gráfico de dados reais
plt.plot(mes, A, color=cor, label='Previsão')  # Gráfico de previsão (A) com cor cornflowerblue

# Configurações de título, eixos e rótulos
plt.title('Valor médio da previsão clássica da temperatura', fontsize=18)
plt.axis([0, 15, 26, 29])  # Ajuste para o eixo Y de acordo com os dados
plt.xlabel(r'Tempo (dias)', fontsize=16)
plt.ylabel('Temperatura (°C)', fontsize=16)

# Adicionando a legenda
plt.legend()

# Ativando o fundo quadriculado com linhas claras
plt.grid(True, color='lightgray', alpha=0.5)  # Linha tracejada e cor clara

# Salvando o gráfico em um arquivo
plt.savefig('q_prevtemp')

# Exibindo o gráfico
plt.show()
```
